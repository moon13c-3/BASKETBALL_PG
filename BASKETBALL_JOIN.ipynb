```python


dbutils.widgets.text("source_table", "lakehouse.pipeline.basketball")
source_table = dbutils.widgets.get("source_table")

df = spark.table(source_table)

dup_exists = df.count() > df.drop('_gdrive_metadata', '_file_metadata').dropDuplicates().count()

dbutils.jobs.taskValues.set(key = "has_duplicates", value = dup_exists)
```

```python
spark.sql(f"""CREATE OR REPLACE TABLE lakehouse.pipeline.silver_basketball AS 
          SELECT c.season, cast(c.age as bigint), c.team, cast(c.g as bigint) as Games, cast(c.FG as double), cast(c.FGA as double), cast(c.`FG%` as double), cast(c.`3P` as double), cast(c.`3PA` as double), cast(c.`3P%` as double), cast(c.TRB as double), cast(c.AST as double), cast(c.STL as double), cast(c.BLK as double), cast(c.PTS as double), cast(a.PER as double), cast(a.`TS%` as double), cast(a.WS as double) FROM {source_table} c INNER JOIN lakehouse.pipeline.pg2 a ON c.AGE = a.AGE""")
```
